```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/drive/MyDrive/Market_Basket_Optimisation.csv', header=None)
df
```

```python
transactions=[]
for i in range(0,len(df)):
  transactions.append([str(df.values[i,j])for j in range(0,20) if str(df.values[i,j])!='NaN'])
```

```python
def eclat(transactions, min_support):

    item_tid = defaultdict(set)

    # Build vertical format
    for tid, transaction in enumerate(transactions):
        for item in transaction:
            item_tid[item].add(tid)

    def recursive(prefix, items):
        results = []
        items = list(items.items())

        for i in range(len(items)):
            item_i, tid_i = items[i]

            new_prefix = prefix + [item_i]
            support = len(tid_i)

            if support >= min_support:
                results.append((new_prefix, support))

                suffix = {}

                for j in range(i + 1, len(items)):
                    item_j, tid_j = items[j]
                    intersection = tid_i & tid_j

                    if len(intersection) >= min_support:
                        suffix[item_j] = intersection

                results.extend(recursive(new_prefix, suffix))

        return results

    return recursive([], item_tid)
```

```python
from collections import defaultdict
min_support = 2

results = eclat(transactions, min_support)

for itemset, support in results:
    print(itemset, support)
```
